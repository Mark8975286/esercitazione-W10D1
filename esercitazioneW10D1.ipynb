# Esercizio 1 · Array NumPy (semplice · 20 minuti)

## Esercizio 1.1
**Domanda:** importare NumPy con l'alias standard `np` e creare un array a partire dalla lista `temperature = [18, 22, 19, 25, 21]` (gradi Celsius).

```python
import numpy as np

temperature = [18, 22, 19, 25, 21]
temp_celsius = np.array(temperature)

print(temp_celsius)
print(type(temperature), "->", type(temp_celsius))
print("Tipo degli elementi:", temp_celsius.dtype)
```

## Esercizio 1.2
**Domanda:** convertire l'array in gradi Fahrenheit applicando la formula `(temperatura * 9/5) + 32` in un'unica operazione vettoriale, senza ciclo for.

```python
temp_fahrenheit = (temp_celsius * 9 / 5) + 32

print("Celsius:   ", temp_celsius)
print("Fahrenheit:", temp_fahrenheit)
```

## Esercizio 1.3
**Domanda:** calcolare media, minimo e massimo dell'array originale in gradi Celsius, usando i metodi statistici di NumPy.

```python
print("Media:  ", temp_celsius.mean(), "°C")
print("Minimo: ", temp_celsius.min(), "°C")
print("Massimo:", temp_celsius.max(), "°C")
```

### Definizione richiesta
**Array NumPy:** struttura che memorizza elementi dello stesso tipo in una griglia (a una o più dimensioni), sulla quale si eseguono operazioni vettoriali.

**Operazione vettoriale:** operazione scritta una sola volta e applicata automaticamente a **tutti gli elementi** dell'array.

**Perché la conversione non richiede un ciclo:** NumPy applica `* 9 / 5 + 32` elemento per elemento in modo interno e ottimizzato, quindi non serve scrivere un `for`; con una lista Python invece `lista * 9` ripeterebbe la lista 9 volte invece di moltiplicare i valori.

# Esercizio 2 · Pandas Series (semplice · 20 minuti)

## Esercizio 2.1
**Domanda:** importare Pandas con l'alias standard `pd` e creare una Series con le popolazioni come valori e i nomi delle città come indice.

```python
import pandas as pd

citta = ["Roma", "Milano", "Napoli", "Torino", "Genova"]
popolazioni = [2750000, 1370000, 910000, 845000, 560000]   # valori indicativi

serie_popolazione = pd.Series(popolazioni, index=citta)

print(serie_popolazione)
print(type(serie_popolazione))
```

## Esercizio 2.2
**Domanda:** accedere al valore di una città specifica usando la sua etichetta di indice, non la posizione numerica.

```python
popolazione_torino = serie_popolazione["Torino"]
print("Popolazione di Torino (per etichetta):", popolazione_torino)

# Stesso risultato con .loc[], l'accesso esplicito per etichetta
print("Con .loc['Torino']:", serie_popolazione.loc["Torino"])
```

## Esercizio 2.3
**Domanda:** creare una seconda Series a partire da un dizionario città:popolazione e verificare che il risultato sia equivalente a quello del punto 1.

```python
dizionario_popolazione = {
    "Roma": 2750000,
    "Milano": 1370000,
    "Napoli": 910000,
    "Torino": 845000,
    "Genova": 560000,
}
serie_da_dizionario = pd.Series(dizionario_popolazione)

print(serie_da_dizionario)
print()
print("Le due Series sono equivalenti?", serie_popolazione.equals(serie_da_dizionario))
```

### Definizione richiesta
**Series:** array monodimensionale etichettato, il tipo di dato base di Pandas: ogni valore è associato a un'etichetta dell'indice.

**Differenza tra indice di una Series e posizione in una lista:** in una lista l'elemento si identifica solo con la sua posizione numerica (0, 1, 2…), che cambia se la lista viene riordinata; in una Series ogni valore ha un'**etichetta** (qui il nome della città) che resta legata al valore anche se l'ordine cambia, quindi `serie["Torino"]` restituisce sempre la popolazione di Torino.

# Esercizio 3 · Pandas DataFrame (semplice · 20 minuti)

## Esercizio 3.1
**Domanda:** costruire un dizionario con le tre liste (nome, categoria, prezzo) e creare un DataFrame con `pd.DataFrame()`.

```python
catalogo = {
    "nome": ["Penna", "Quaderno", "Zaino", "Calcolatrice"],
    "categoria": ["Cancelleria", "Cancelleria", "Accessori", "Elettronica"],
    "prezzo": [1.50, 3.00, 25.00, 12.90],
}

df_catalogo = pd.DataFrame(catalogo)

print(df_catalogo)
print(type(df_catalogo))
```

## Esercizio 3.2
**Domanda:** ispezionare il DataFrame con `.shape` e `.columns`, stampando entrambi i risultati.

```python
print("Shape (righe, colonne):", df_catalogo.shape)
print("Colonne:", df_catalogo.columns)
print("Colonne come lista:", list(df_catalogo.columns))
```

## Esercizio 3.3
**Domanda:** selezionare la colonna prezzo del DataFrame e verificare con `type()` che il risultato sia una Series.

```python
colonna_prezzo = df_catalogo["prezzo"]
print(colonna_prezzo)
print("Tipo con parentesi singole:", type(colonna_prezzo))

# Confronto con le doppie parentesi quadre
colonna_prezzo_df = df_catalogo[["prezzo"]]
print("Tipo con parentesi doppie: ", type(colonna_prezzo_df))
```

**Nota:** `df["colonna"]` restituisce una **Series**; `df[["colonna"]]`, con doppie parentesi quadre, restituisce invece un **DataFrame** di una sola colonna.

### Definizione richiesta
**DataFrame:** struttura tabellare bidimensionale con righe e colonne etichettate, costruita da più Series che condividono lo stesso indice.

**Perché ogni colonna è una Series:** un DataFrame è di fatto un insieme di Series affiancate, una per colonna, tutte allineate sullo stesso indice di riga; selezionando una colonna si estrae quindi una di queste Series, con i suoi valori e l'indice delle righe.

# Esercizio 4 · Lettura e ispezione (avanzato · 30 minuti)

## Esercizio 4.1
**Domanda:** costruire il DataFrame `ordini` a partire da un dizionario di liste con le colonne `id_ordine`, `cliente`, `importo`, `citta` e 12 righe, inserendo almeno due valori mancanti (`None`) nella colonna importo.

```python
dati_ordini = {
    "id_ordine": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112],
    "cliente": ["Rossi", "Bianchi", "Verdi", "Russo", "Ferrari", "Esposito",
                "Romano", "Colombo", "Ricci", "Marino", "Greco", "Bruno"],
    "importo": [120.50, 89.90, None, 250.00, 45.00, 310.75,
                None, 99.99, 175.20, None, 60.00, 420.00],
    "citta": ["Torino", "Milano", "Torino", "Roma", "Milano", "Torino",
              "Napoli", "Roma", "Torino", "Milano", "Napoli", "Torino"],
}

ordini = pd.DataFrame(dati_ordini)

print(ordini)
print()
print("Shape:", ordini.shape)
print("Valori None inseriti in importo:", dati_ordini["importo"].count(None))
```

## Esercizio 4.2
**Domanda:** usare `.info()` per individuare quante righe della colonna importo contengono valori non nulli.

```python
ordini.info()
```

```python
# Verifica del dato letto in .info()
non_nulli = ordini["importo"].notna().sum()
mancanti = ordini["importo"].isna().sum()
print("Valori non nulli in importo:", non_nulli)
print("Valori mancanti in importo: ", mancanti)
print("Totale righe:", non_nulli + mancanti)
```

**Risultato:** `.info()` riporta **9 non-null** su 12 righe per la colonna `importo`, che corrisponde ai dati inseriti (12 righe − 3 `None`). I `None` sono stati convertiti da Pandas in `NaN`.

## Esercizio 4.3
**Domanda:** usare `.describe()` sulla colonna importo e `.value_counts()` sulla colonna citta, confrontando le informazioni restituite dai due metodi.

```python
print("=== describe() su importo ===")
print(ordini["importo"].describe())
```

```python
print("=== value_counts() su citta ===")
print(ordini["citta"].value_counts())
```

**Confronto:** `.describe()` riassume una colonna **numerica** con statistiche (conteggio dei non nulli, media, deviazione standard, minimo, quartili, massimo); `.value_counts()` lavora su una colonna **categorica** e conta quante volte compare ciascun valore distinto (qui quanti ordini per città).

### Definizione richiesta
**Valore mancante:** cella che non contiene un dato; in Pandas è rappresentata da `NaN` (o `None`/`NA`) e viene esclusa dai calcoli statistici.

**Differenza tra `.info()` e `.describe()`:** `.info()` descrive la **struttura** del DataFrame (numero di righe, nomi delle colonne, tipo di dato e numero di valori non nulli per colonna); `.describe()` descrive il **contenuto** delle colonne numeriche con statistiche riassuntive.

# Esercizio 5 · .loc[] e .iloc[] (avanzato · 30 minuti)

## Esercizio 5.1
**Domanda:** impostare la colonna `id_ordine` come indice del DataFrame con `.set_index()`, assegnando il risultato a una nuova variabile.

```python
ordini_idx = ordini.set_index("id_ordine")

print(ordini_idx)
print()
print("Indice del nuovo DataFrame:", list(ordini_idx.index))
print("Colonne del DataFrame originale (invariato):", list(ordini.columns))
```

## Esercizio 5.2
**Domanda:** sul DataFrame con il nuovo indice, selezionare una riga con `.loc[]` usando un'etichetta di `id_ordine`, e la stessa riga con `.iloc[]` usando la sua posizione numerica.

```python
riga_loc = ordini_idx.loc[104]    # etichetta id_ordine = 104
riga_iloc = ordini_idx.iloc[3]    # posizione 3 (quarta riga, si parte da 0)

print("Con .loc[104]:")
print(riga_loc)
print()
print("Con .iloc[3]:")
print(riga_iloc)
print()
print("Le due righe sono identiche?", riga_loc.equals(riga_iloc))
```

## Esercizio 5.3
**Domanda:** selezionare un intervallo di tre righe consecutive sia con `.loc[]` sia con `.iloc[]`, confrontando quante righe restituisce ciascuno dei due intervalli.

```python
intervallo_loc = ordini_idx.loc[103:105]    # etichette: estremo finale INCLUSO
intervallo_iloc = ordini_idx.iloc[2:5]      # posizioni: estremo finale ESCLUSO

print(".loc[103:105] ->", len(intervallo_loc), "righe")
print(intervallo_loc)
print()
print(".iloc[2:5] ->", len(intervallo_iloc), "righe")
print(intervallo_iloc)
```

```python
# Stessi numeri di inizio e fine: il risultato cambia
print(".loc[103:105]  ->", len(ordini_idx.loc[103:105]), "righe (105 incluso)")
print(".iloc[2:4]     ->", len(ordini_idx.iloc[2:4]), "righe (posizione 4 esclusa)")
```

**Risultato:** per ottenere le stesse 3 righe servono `.loc[103:105]` e `.iloc[2:5]`; con estremi equivalenti (`.loc[103:105]` contro `.iloc[2:4]`) `.loc` restituisce **3** righe e `.iloc` solo **2**.

### Definizione richiesta
- **Indicizzazione per etichetta (`.loc[]`):** seleziona righe e colonne usando i **nomi** dell'indice e delle colonne.
- **Indicizzazione per posizione (`.iloc[]`):** seleziona righe e colonne usando la loro **posizione numerica**, partendo da 0.

**Perché il numero di righe può differire:** nello slicing `.loc[]` **include** l'etichetta finale, mentre `.iloc[]` segue la regola di Python ed **esclude** la posizione finale; a parità di estremi, `.loc` restituisce quindi una riga in più.

# Esercizio 6 · Masking (avanzato · 30 minuti)

## Esercizio 6.1
**Domanda:** costruire una maschera booleana che individui gli ordini con importo superiore a una soglia a scelta, e usarla per filtrare il DataFrame con `.loc[]`.

```python
soglia = 100

maschera_importo = ordini["importo"] > soglia
print("Maschera booleana:")
print(maschera_importo)
print(type(maschera_importo), "- lunghezza:", len(maschera_importo))
```

```python
ordini_sopra_soglia = ordini.loc[maschera_importo]

print(f"Ordini con importo > {soglia} €:", len(ordini_sopra_soglia), "righe")
print(ordini_sopra_soglia)
```

## Esercizio 6.2
**Domanda:** costruire un filtro che combini due condizioni con l'operatore `&`: importo superiore alla soglia scelta al punto 1 e citta uguale a una città a scelta.

```python
citta_scelta = "Torino"

filtro_and = (ordini["importo"] > soglia) & (ordini["citta"] == citta_scelta)
risultato_and = ordini.loc[filtro_and]

print(f"Ordini con importo > {soglia} € E città = {citta_scelta}:", len(risultato_and), "righe")
print(risultato_and)
```

## Esercizio 6.3
**Domanda:** costruire un filtro che usi l'operatore `|` per selezionare gli ordini di due città diverse, poi negare lo stesso filtro con l'operatore `~` e confrontare il numero di righe dei due risultati.

```python
filtro_or = (ordini["citta"] == "Torino") | (ordini["citta"] == "Milano")

risultato_or = ordini.loc[filtro_or]
risultato_not = ordini.loc[~filtro_or]

print("Ordini di Torino OPPURE Milano:", len(risultato_or), "righe")
print(risultato_or)
print()
print("Ordini NON di Torino né Milano (~):", len(risultato_not), "righe")
print(risultato_not)
```

```python
totale_righe = len(ordini)
somma = len(risultato_or) + len(risultato_not)

print("Righe filtro |:", len(risultato_or))
print("Righe filtro ~:", len(risultato_not))
print("Somma:", somma, "| Righe totali del DataFrame:", totale_righe)
print("La somma corrisponde al totale?", somma == totale_righe)
```

**Nota:** ogni condizione va racchiusa tra parentesi tonde, perché `&` e `|` hanno precedenza più alta di `>` e `==`.

### Definizione richiesta
**Masking:** tecnica che applica una condizione a una colonna ottenendo una Series di booleani (`True`/`False`) con lo stesso indice del DataFrame, usata per tenere solo le righe in cui la condizione è vera.

**Perché si usano `&` e `|` e non `and`/`or`:** `and` e `or` valutano un oggetto intero come un singolo `True` o `False`, e su una Series con molti valori Python non sa quale scegliere (genera un `ValueError`); `&`, `|` e `~` invece lavorano **elemento per elemento**, confrontando riga per riga le due maschere.

```python
# Dimostrazione: and su due Series genera un errore
try:
    (ordini["importo"] > soglia) and (ordini["citta"] == "Torino")
except ValueError as errore:
    print("Tipo di eccezione:", type(errore).__name__)
    print("Messaggio:", errore)
```
